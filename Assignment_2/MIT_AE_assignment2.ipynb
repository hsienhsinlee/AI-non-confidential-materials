```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Separate features (X) and target (y)
X = loandata.drop('not.fully.paid', axis=1)
y = loandata['not.fully.paid']

# Handle categorical variables using one-hot encoding
X = pd.get_dummies(X, columns=['purpose'], drop_first=True)
```

```python
# Split the data into training and testing sets
# 70% for training, 30% for testing, with random_state=144 for reproducibility
# stratify=y ensures that the proportion of target variable is the same in both train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=144, stratify=y)

# Create the 'train' and 'test' DataFrames by concatenating features and target
train = pd.concat([X_train, y_train], axis=1)
test = pd.concat([X_test, y_test], axis=1)

print("Training set shape:", train.shape)
print("Testing set shape:", test.shape)
```

```python
# Initialize and train the Logistic Regression model
# Using default solver for now, can be optimized later if needed
model = LogisticRegression(random_state=144, max_iter=1000)
model.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
# You can inspect the model's coefficients or make predictions next.
```
